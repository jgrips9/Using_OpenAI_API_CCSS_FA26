```python
#Perform a search based on the results that I get. Look for specific characters.
companies = ["Nike", "Disney", "Amazon"]
results = []
for i in companies:
    query = "What is the total networth of the comany " + i
    response = client.responses.create(
    model="gpt-4.1",
    tools=[{"type": "web_search_preview"}],
    input=query
)
    responses = response.to_dict()
    results.append(responses['output'][1]['content'][0]['text'])
```

```python
#Create dataset.
import pandas
networth_companies = pandas.DataFrame({'company_name': companies, 'ChatGPT Response': results})
display(networth_companies)
```

```python
import re

def extract_dollar_amount_and_text(text):
    # Make sure we have a string
    if not isinstance(text, str):
        return None, None

    # Find a dollar amount, allowing commas and optional decimals
    pattern = re.compile(r'\$([\d,]+(?:\.\d+)?)\s*(.*)', re.DOTALL)
    match = pattern.search(text)

    if match:
        # Remove commas before converting to float
        amount_str = match.group(1).replace(',', '')
        amount_float = float(amount_str)

        text_after = match.group(2).strip()

        return amount_float, text_after

    return None, None
```

```python
dollar_amt_num = []
dollar_amt_text = []

for i in networth_companies['ChatGPT Response']:

    amount, output_text = extract_dollar_amount_and_text(i)

    dollar_amt_num.append(amount)

    if output_text:
        texts = re.split(r'(?<=\.)\s*', output_text)
        dollar_amt_text.append(texts[0] if texts else '')
    else:
        dollar_amt_text.append('')
```

```python
networth_companies['amount'] = dollar_amt_num
networth_companies['scale'] = dollar_amt_text
```

```python
display(networth_companies)
```

```python
#Import spreadsheet.
NE_States = pandas.read_excel('NE States.xlsx')
display(NE_States)
```

```python
counties = []
city = []
for i in NE_States['state']:
  query1 = "How many counties are present in the state named: " + i
  query2= "What is the largest city in this state?"
  response = client.responses.create(
    model="gpt-4.1",
    tools=[{"type": "web_search_preview"}],
    input=query1
)
  counties.append(response.output_text)
  conversation_id = response.id
  follow_up = client.responses.create(
    model="gpt-4.1",
    input=query2,
    #This allows the conversation to be carried on.
    previous_response_id=conversation_id
)
  city.append(follow_up.output_text)
```

```python
NE_States['counites'] = counties
NE_States['city'] = city
```
